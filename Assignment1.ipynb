Madhava Pi Calculation

```python
def the_pi(N):
    pi=0
    for n in range(1,N):
        if n == 1:
            term = 1
        else:
            term = -((-1)**n) / (2*n - 1)
        pi = pi + term
    
    return 4 * pi
        
```

```python
for N in [100, 1000, 10000]:
    print(f"N={N} = pi = {the_pi(N):2.7f}")
```

```python
import math 
n_val = [10,100,2000,3000,40000,500000]
abs_err=[]

for N in n_val:
    error = abs(the_pi(N)-math.pi)
    abs_err.append(error)
```

```python
import matplotlib.pyplot as plt
plt.plot(n_val, abs_err)
plt.xscale('log')
plt.yscale('log')
plt.xlabel("Num of terms")
plt.ylabel("Error")
plt.title("Convergence of the Madhava-Leibniz series for pi")
plt.show()
```

```python
for N, error in zip(n_val, abs_err):
    print("N = ", N, "error = ", error)
```

Reasonable accuracy: 
At N= 40000, it was reasonable accuracy, as the error was 2.5000625013316125e-05, and when the N is increasing the error is also increasing.
